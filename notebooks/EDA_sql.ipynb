```python
import duckdb
# create a connection to duckdb
con = duckdb.connect()
print("DuckDB SQL Engine Ready!")
```

```python
# read raw data
view_query = """
    SELECT * FROM read_csv_auto('../data/SGJobData.csv')
    LIMIT 4;
    """
# show table
con.sql(view_query).df()
```

```python
# try to load raw data from hard drive to temporary table in RAM to work on
setup_query = """
    CREATE OR REPLACE TABLE raw_jobs AS
    SELECT * FROM read_csv_auto('../data/SGJobData.csv');
"""
con.sql(setup_query)
```

```python
# view raw_Job table
con.sql("SELECT * FROM raw_jobs LIMIT 4;").df()
```

```python
# describe table
con.sql("DESCRIBE SELECT * FROM raw_Jobs;")
```

```python
con.sql("DESCRIBE raw_Jobs;").df()
```

```python
# shows Data Types, Min, Max, Unique Counts, and Null percentages for ALL columns
con.sql("SUMMARIZE raw_jobs;").df()
```

```python
# futher check on salary_type
unique_query = """
    SELECT DISTINCT salary_type, count(*) as count
    FROM raw_jobs
    GROUP BY salary_type;
"""

con.sql(unique_query).df()
```

```python
# show value count for all column (same as df.info or value_count in pandas)
# Step 1: Tell DuckDB to get a list of all column names in the 'raw_jobs' table
columns_df = con.sql("SELECT column_name FROM information_schema.columns WHERE table_name = 'raw_jobs'").df()

# Step 2: Extract the column names as a Python list
col_list = columns_df['column_name'].tolist()

# Step 3: Loop through each column and run a value_counts SQL query!
for col in col_list:
    print(f"\n--- Top 5 Value Counts for: {col} ---")
    
    # We use Python's f-string to inject the column name into the SQL query
    query = f"""
        SELECT {col}, COUNT(*) AS frequency
        FROM raw_jobs
        GROUP BY {col}
        ORDER BY frequency DESC
        LIMIT 5;
    """
    
    # Using display() makes the output look like nice pandas dataframes in Jupyter
    display(con.sql(query).df())
```

```python
# rename all the column (as per Data Dictionary)
rename_query = """
    CREATE OR REPLACE TABLE renamed_jobs AS 
    SELECT 
        -- SYNTAX: original_column_name AS New_Column_Name
        
        categories AS Category,
        employmentTypes AS Employment_Type,
        positionLevels AS Position_Level,
        postedCompany_name AS Company_Name,
        status_jobStatus AS Job_Status,
        
        metadata_expiryDate AS Expiry_Date,
        metadata_newPostingDate AS New_Post_Date,
        metadata_originalPostingDate AS Orig_Post_Date,
        
        metadata_isPostedOnBehalf AS Posted_On_Behalf,
        metadata_jobPostId AS Job_Post_ID,
        
        metadata_repostCount AS Repost_Date,
        metadata_totalNumberJobApplication AS Total_Applications,
        metadata_totalNumberOfView AS Total_Views,
        minimumYearsExperience AS Min_Yrs_Experience,
        numberOfVacancies AS Vacancies,
        salary_minimum AS Salary_Min,
        salary_maximum AS Salary_Max
        
        -- drop "salary_type" at the same time, as it has only 1 unique value 'monthly'
        -- drop "status_id", as it has only 1 unique value '0'
        
    FROM raw_jobs;
"""

# Execute the query
con.sql(rename_query)

# View the result to verify the column headers have changed
con.sql("SELECT * FROM renamed_jobs LIMIT 5").df()
```

```python
# cleaning query
transform_query = """
    COPY (
        SELECT 
            -- 1. Explode JSON
            json_extract_string(categories, '$[0].category') AS Category,
            
            -- 2. Rename standard columns
            employmentTypes AS Employment_Type,
            positionLevels AS Position_Level,
            postedCompany_name AS Company_Name,
            status_jobStatus AS Job_Status,
            
            -- 3. Date formats
            CAST(metadata_expiryDate AS DATE) AS Expiry_Date,
            CAST(metadata_newPostingDate AS DATE) AS New_Post_Date,
            CAST(metadata_originalPostingDate AS DATE) AS Orig_Post_Date,
            
            -- 4. Booleans
            CAST(metadata_isPostedOnBehalf AS BOOLEAN) AS Posted_On_Behalf,
            
            -- 5. IDs
            metadata_jobPostId AS Job_Post_ID,
            
            -- 6. Numerics
            CAST(metadata_repostCount AS INTEGER) AS Repost_Count, 
            CAST(metadata_totalNumberJobApplication AS INTEGER) AS Total_Applications,
            CAST(metadata_totalNumberOfView AS INTEGER) AS Total_Views,
            CAST(minimumYearsExperience AS INTEGER) AS Min_Yrs_Experience,
            CAST(numberOfVacancies AS INTEGER) AS Vacancies,
            CAST(salary_minimum AS INTEGER) AS Salary_Min,
            CAST(salary_maximum AS INTEGER) AS Salary_Max
            
        FROM raw_jobs
        
        -- Filter out bad rows
        WHERE metadata_jobPostId IS NOT NULL 
          AND metadata_jobPostId != ''
          
    ) TO '../data/SGJobData_Cleaned_SQL.csv' (HEADER, DELIMITER ',');
"""

print("Executing SQL Transformation...")

con.sql(transform_query)
print("Success! Cleaned data saved to 'data/SGJobData_Cleaned_SQL.csv'")
```
