```python
# Check missing values in category: (isna and isnull are the sam haha)
print(df["categories"].isna().sum())  
```

```python
# import json library to deal with JSON data in the "categories" column:
import json
```

```python
# Parse the JSON array and extract the value of the "category" names in the "categories" column into a new column "category_single":
df["category_single"] = df["categories"].apply(lambda x: [item["category"] for item in json.loads(x)] if isinstance(x, str) else [])
```

```python
# Check value of the sample row in the new column "category_single":
print(df["category_single"].iloc[0])
```

```python
# Explode the lists into separate rows so each category has its own row.
# This is important as it adding new row for each category in the list, which will allow us to analyze the categories more effectively.
df_temp = df.explode("category_single")
```

```python
# check the value of the new exploded dataframe: top 20 only
df_temp["category_single"].value_counts().head(20)
```

```python
# Describe the exploded new dataframe: df_temp
df_temp["category_single"].describe()
```

```python
# check the shape of the exploded dataframe: Now we have : 1771817-1048585= 723232 new rows and 1 column added (category_single)
# All the data remains the same, if there are more than 1 category in the list, it will be added as a new separate rows right after the original row.
# So the total number of rows will be increased by the number of categories in the list.
df_temp.shape
```

```python
# Save the exploded dataframe to a new CSV file:
# Get the output file at the same data folder
# You use DbGate to view the CSV file.

df_temp.to_csv("../data/output1.csv", index=True, sep=";")
```
