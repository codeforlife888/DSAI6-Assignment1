```python
#load libraries
import pandas as pd
import numpy as np
```

Read csv file to data frame

```python
#load data:
df = pd.read_csv("../data/SGJobData.csv")  
```

```python
# check data size: (Original data: 1048585 rows, 22 columns), 
df.shape
```

```python
# view the data
df
```

```python
# check data types and count of non-null values
df.info()
```

```python
# drop duplicates where all rows are the same
# (Original data: 1048585 rows, 22 columns)
df = df.drop_duplicates()
```

```python
df.info()
```

```python
# view the dataframe and check the data types and count of non-null values
df.describe(include="all").round(1)
```

```python
# check on "occupationID" column:
df["occupationId"].value_counts(dropna=False)
```

```python
# drop column "ocupationId as all values are null
df = df.drop(columns=["occupationId"])  
```

```python
# Describe 1 test column:
df["salary_maximum"].describe().round(0)
```

```python
# Describe "categories":
df["categories"].describe()
```

```python
# View Json data in the "categories" column:
print(df["categories"].head())
```

```python
# View the Json raw data in the "categories" column to understand the structure of the data:
print(df["categories"][0])  
```

```python
# import json library to deal with JSON data in the "categories" column:
import json
```

```python
# Parse the JSON array and extract the value of the "category" names in the "categories" column into a new column "category_single":
df["category_single"] = df["categories"].apply(lambda x: [item["category"] for item in json.loads(x)] if isinstance(x, str) else [])
```

```python
# Check value of the sample row in the new column "category_single":
print(df["category_single"].iloc[0])
```

```python
# Explode the lists into separate rows so each category has its own row.
# This is important as it adding new row for each category in the list, which will allow us to analyze the categories more effectively.
df_exploded = df.explode("category_single")
```

```python
# check the value of the new exploded dataframe: top 20 only
df_exploded["category_single"].value_counts().head(20)
```

```python
df_exploded.info()
```

```python
# Check the value of the new exploded dataframe: top 40 only
df_exploded["category_single"].value_counts().head(40).plot(kind="barh", figsize=(10, 8), title="Top 40 Categories")
```

```python
# check for null values in each column
df_exploded.isnull().sum()
```

```python
# check if the "metadata_jobPostId" column is unique and if there are any null values in this column
print(df_exploded["metadata_jobPostId"].is_unique)  # check if the column is unique
print(df_exploded["metadata_jobPostId"].isnull().sum())  # check for null values
```

```python
# find the null values in the "metadata_jobPostId" column and view the corresponding rows in the dataframe
df_exploded[df_exploded["metadata_jobPostId"].isnull()]
```

```python
# drop this row as it is not useful for our analysis
df_exploded = df_exploded.dropna(subset=["metadata_jobPostId"])
```

```python
# describe metadata_jobPostId column after dropping the null value row
df_exploded["metadata_jobPostId"].describe()
```

```python
# drop "categories" column as we have already extracted the category names into a new column "category_single"
df_exploded = df_exploded.drop(columns=["categories"])
```

```python
# view the final exploded dataframe and check the data types and count of non-null values
df_exploded.info()
```

```python
# rename columns name:
df_exploded = df_exploded.rename(columns={"employmentTypes": "Employment_Types",
                        "metadata_expiryDate": "Expiry_Date",
                        "metadata_isPostedOnBehalf" : "Posted_On_Behalf",
                        "metadata_jobPostId": "Job_Post_ID",
                        "metadata_newPostingDate": "New_Post_Date",  
                        "metadata_originalPostingDate": "Orig_Post_Date",
                        "metadata_repostCount": "Repost_Count",
                        "metadata_totalNumberJobApplication" : "Total_Applications",                     
                        "metadata_totalNumberOfView": "Total_Views",
                        "minimumYearsExperience": "Min_Yrs_Experience",
                        "numberOfVacancies": "Vacancies",
                        "occupationId": "Occupation_ID",
                        "positionLevels": "Position_Level",
                        "postedCompany_name": "Company_Name",
                        "salary_maximum" : "Salary_Max",
                        "salary_minimum" : "Salary_Min",
                        "salary_type": "Salary_Type",
                        "status_id": "Status_ID",
                        "status_jobStatus": "Job_Status",
                        "title": "Job_Title",
                        "average_salary": "Avg_Salary",
                        "category_single": "Category",
                     }) 
```
